## 2. Prepare and clean the data
This section preserves the original data in `df_raw`, standardizes text labels, converts numeric fields safely, and checks missing values and duplicates. We do not automatically remove duplicates because repeated rows should be investigated before deletion.

```python
# Make a backup and clean text formatting
if "df" in globals():
    df_raw = df.copy()
    # Trim whitespace from text values
    for col in df.select_dtypes(include=["object", "string"]).columns:
        df[col] = df[col].astype("string").str.strip()
    print("Backup saved as df_raw. Leading/trailing spaces trimmed.")
```

```python
# # Normalize common category variants in Item Fat Content
if "df" in globals():
    if "Item Fat Content" in df.columns:
        df["Item Fat Content"] = (
            df["Item Fat Content"].astype("string").str.strip().str.lower()
            .replace({
                "lf": "low fat",
                "low fat": "low fat",
                "reg": "regular",
                "regular": "regular"
            })
            .str.title()
        )
        display(df["Item Fat Content"].value_counts(dropna=False).to_frame("Count"))
```

```python
#Missing values and duplicate checks
if "df" in globals():
    missing = df.isna().sum().sort_values(ascending=False)
    missing_pct = (df.isna().mean() * 100).sort_values(ascending=False)
    missing_report = pd.DataFrame({"Missing Count": missing, "Missing %": missing_pct})
    display(missing_report[missing_report["Missing Count"] > 0])
    print("Exact duplicate rows:", int(df.duplicated().sum()))
    print("Duplicate Item Identifier values:",
          int(df["Item Identifier"].duplicated().sum()) if "Item Identifier" in df.columns else "Column unavailable")
```

```python
# Convert expected numeric columns and inspect unusual values
if "df" in globals():
    numeric_candidates = [
        "Outlet Establishment Year", "Item Visibility", "Item Weight", "Sales", "Rating"
    ]
    for col in numeric_candidates:
        if col in df.columns:
            df[col] = pd.to_numeric(df[col], errors="coerce")

    display(df[[c for c in numeric_candidates if c in df.columns]].describe().T)
    print("Rows with negative sales:",
          int((df["Sales"] < 0).sum()) if "Sales" in df.columns else "Sales column unavailable")
    print("Rows with negative item visibility:",
          int((df["Item Visibility"] < 0).sum()) if "Item Visibility" in df.columns else "Column unavailable")
    if "Rating" in df.columns:
        print("Rating values outside 1–5:",
              int((~df["Rating"].between(1, 5) & df["Rating"].notna()).sum()))
```

### Handling missing values
The code below uses the median for `Item Weight` and the mode for categorical columns only as a practical analysis option. Missing `Sales` values are not imputed because sales are the main outcome; rows without sales cannot contribute reliably to sales analysis. Review the printed counts before deciding whether to keep or remove any records.

```python
# Cell 8 — Conservative handling of missing values
if "df" in globals():
    if "Item Weight" in df.columns and df["Item Weight"].isna().any():
        df["Item Weight"] = df["Item Weight"].fillna(df["Item Weight"].median())

    # Fill missing categories with the most frequent category (mode), if one exists
    for col in df.select_dtypes(include=["object", "string"]).columns:
        if df[col].isna().any():
            modes = df[col].mode(dropna=True)
            if not modes.empty:
                df[col] = df[col].fillna(modes.iloc[0])

    # Exclude records without the target variable for analyses requiring sales
    if "Sales" in df.columns:
        sales_df = df[df["Sales"].notna()].copy()
    else:
        sales_df = df.copy()

    # print("Rows in cleaned dataset:", len(df))
    print("Rows available for sales analysis:", len(sales_df))
    print("Remaining missing values:")
    display(df.isna().sum().to_frame("Missing Count").query("`Missing Count` > 0"))
```

```python
#Product and outlet variety
if "df" in globals():
    for col in ["Item Type", "Outlet Type", "Outlet Location Type", "Outlet Size", "Item Fat Content"]:
        if col in df.columns:
            print(f"\n{col}: {df[col].nunique(dropna=True)} distinct values")
            display(df[col].value_counts(dropna=False).to_frame("Record Count"))
```

```python
#Sales by product category
if "Sales" in sales_df.columns and "Item Type" in sales_df.columns:
    sales_by_item_type = (sales_df.groupby("Item Type", dropna=False)["Sales"]
                          .agg(["sum", "mean", "count"])
                          .sort_values("sum", ascending=False))
    sales_by_item_type.columns = ["Total Sales", "Average Sales per Record", "Record Count"]
    display(sales_by_item_type)

    ax = sales_by_item_type["Total Sales"].sort_values().plot(kind="barh", figsize=(10, 7))
    ax.set_title("Total Sales by Item Type")
    ax.set_xlabel("Total Sales")
    ax.set_ylabel("Item Type")
    plt.tight_layout()
    plt.show()
```

```python
# Sales by outlet dimensions
if "Sales" in sales_df.columns:
    for col in ["Outlet Type", "Outlet Location Type", "Outlet Size", "Outlet Establishment Year"]:
        if col in sales_df.columns:
            summary = (sales_df.groupby(col, dropna=False)["Sales"]
                       .agg(["sum", "mean", "count"]).sort_values("sum", ascending=False))
            summary.columns = ["Total Sales", "Average Sales per Record", "Record Count"]
            print(f"Sales by {col}")
            display(summary)

            if summary.shape[0] <= 20:
                ax = summary["Total Sales"].sort_values().plot(kind="barh", figsize=(9, max(3, 0.4*len(summary))))
                ax.set_title(f"Total Sales by {col}")
                ax.set_xlabel("Total Sales")
                plt.tight_layout()
                plt.show()
```

```python
# Rating overview
if "Rating" in df.columns:
    print("Average rating:", round(df["Rating"].mean(), 3))
    display(df["Rating"].describe().to_frame("Rating statistics"))
    ax = df["Rating"].dropna().plot(kind="hist", bins=10, edgecolor="black", figsize=(8, 4))
    ax.set_title("Distribution of Customer Ratings")
    ax.set_xlabel("Rating")
    plt.tight_layout()
    plt.show()
```

## 4. Objective 2 — Data reliability and quality

```python
# Outlier review using IQR
if "df" in globals():
    for col in ["Sales", "Item Visibility", "Item Weight", "Rating"]:
        if col in df.columns:
            s = df[col].dropna()
            if len(s) > 0:
                q1, q3 = s.quantile([0.25, 0.75])
                iqr = q3 - q1
                lower, upper = q1 - 1.5 * iqr, q3 + 1.5 * iqr
                flagged = ((df[col] < lower) | (df[col] > upper)).sum()
                print(f"{col}: IQR lower={lower:.3f}, upper={upper:.3f}, flagged rows={int(flagged)}")
```

```python
#Consistency checks and duplicate review
if "df" in globals():
    if "Item Fat Content" in df.columns:
        print("Fat-content categories after standardization:")
        display(df["Item Fat Content"].value_counts(dropna=False))
    if "Outlet Establishment Year" in df.columns:
        print("Establishment year range:",
              df["Outlet Establishment Year"].min(), "to", df["Outlet Establishment Year"].max())
    print("Exact duplicate rows:", int(df.duplicated().sum()))
    print("Data types:")
    display(df.dtypes.to_frame("Data Type"))
```

```python
# Item visibility versus sales
if {"Item Visibility", "Sales"}.issubset(sales_df.columns):
    plot_df = sales_df[["Item Visibility", "Sales"]].dropna()
    print("Pearson correlation:", round(plot_df["Item Visibility"].corr(plot_df["Sales"]), 3))
    plt.figure(figsize=(8, 5))
    sns.scatterplot(data=plot_df, x="Item Visibility", y="Sales", alpha=0.35)
    plt.title("Item Visibility vs Sales")
    plt.tight_layout()
    plt.show()
```

```python
#Item weight versus sales
if {"Item Weight", "Sales"}.issubset(sales_df.columns):
    plot_df = sales_df[["Item Weight", "Sales"]].dropna()
    print("Pearson correlation:", round(plot_df["Item Weight"].corr(plot_df["Sales"]), 3))
    plt.figure(figsize=(8, 5))
    sns.scatterplot(data=plot_df, x="Item Weight", y="Sales", alpha=0.35)
    plt.title("Item Weight vs Sales")
    plt.tight_layout()
    plt.show()
```

```python
# Average sales by rating
if {"Rating", "Sales"}.issubset(sales_df.columns):
    rating_sales = sales_df.groupby("Rating", dropna=False)["Sales"].agg(["mean", "sum", "count"]).sort_index()
    rating_sales.columns = ["Average Sales", "Total Sales", "Record Count"]
    display(rating_sales)
    ax = rating_sales["Average Sales"].plot(kind="bar", figsize=(8, 4))
    ax.set_title("Average Sales by Rating")
    ax.set_ylabel("Average Sales")
    plt.tight_layout()
    plt.show()
```

```python
# Compare key categorical groups using mean and total sales
if "Sales" in sales_df.columns:
    group_cols = [c for c in ["Item Fat Content", "Item Type", "Outlet Type",
                              "Outlet Location Type", "Outlet Size"] if c in sales_df.columns]
    for col in group_cols:
        group_summary = sales_df.groupby(col, dropna=False)["Sales"].agg(["sum", "mean", "count"])
        group_summary.columns = ["Total Sales", "Average Sales", "Record Count"]
        group_summary = group_summary.sort_values("Total Sales", ascending=False)
        print(f"Sales comparison: {col}")
        display(group_summary)
```
